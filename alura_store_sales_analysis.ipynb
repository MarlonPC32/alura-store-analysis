# Analysis conclusions

# Final Report — Alura Store Performance Analysis

## Introduction

The goal of this analysis is to evaluate the performance of the four stores in the **Alura Store** chain, to inform a strategic decision about which store could be sold to invest in a new business.

The analysis covers key business indicators per store: **total revenue, best- and worst-selling product categories, average customer ratings, best- and worst-selling products, and average shipping cost**, with visualizations to compare stores and surface relevant patterns.

---

## Revenue analysis

Total revenue per store:

- **Store 1:** 1,150,880,400
- **Store 2:** 1,116,343,500
- **Store 3:** 1,098,019,600
- **Store 4:** 1,038,375,700

**Store 4 has the lowest revenue in the chain**, trailing Store 1 by about 112M. Store 1 leads on revenue.

---

## Sales by category

Best-selling categories per store:

- **Store 1:** Furniture
- **Store 2:** Furniture
- **Store 3:** Furniture
- **Store 4:** Furniture

Least-sold categories per store:

- **Store 1:** Home goods (171)
- **Store 2:** Home goods (181)
- **Store 3:** tie — Home goods (177) and Musical instruments (177)
- **Store 4:** Musical instruments (170)

Note: an earlier version of this report listed Musical instruments alone for Store 3; the counts show a tie at 177 units each.

---

## Customer satisfaction

Average customer rating per store:

- **Store 1:** 3.98
- **Store 2:** 4.04
- **Store 3:** 4.05
- **Store 4:** 4.00

Ratings are stable across stores (all near 4 out of 5). The spread is small — Store 3 rates highest, but the difference is not large enough to drive the decision.

---

## Best- and worst-selling products

Best sellers per store:

- **Store 1:** Microwave
- **Store 2:** "Intro to Programming" (book)
- **Store 3:** Bench kit
- **Store 4:** Box bed

Worst sellers per store:

- **Store 1:** Headphones with microphone
- **Store 2:** Board game
- **Store 3:** Building blocks
- **Store 4:** Electric guitar

---

## Average shipping cost

- **Store 1:** 26,018.61
- **Store 2:** 25,216.24
- **Store 3:** 24,805.68
- **Store 4:** 23,459.46

Store 4 has the lowest average shipping cost. That is an efficiency signal on its own terms.

---

## Conclusion

**Store 4 has the lowest observed revenue of the four stores.** Ratings and shipping costs are similar across the chain, so revenue is the main differentiator in this data.

Whether Store 4 should actually be sold cannot be decided from revenue alone — that requires profitability (margins, fixed costs), store valuation, and growth trajectory, none of which are in this dataset. On the available evidence, Store 4 is the weakest performer by revenue and the natural candidate for a deeper profitability review before any sale decision.
